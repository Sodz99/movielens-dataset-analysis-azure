```
# Connect to Azure Data Lake Storage using SAS Token
spark.conf.set(
    "fs.azure.sas.<Container-name>.<storage-account-name>.blob.core.windows.net",
    "<SAS-Tokens>"
)
```

```
#Access the files in the container
dbutils.fs.ls("abfss://<container-name>@<storage-account-name>.dfs.core.windows.net/")
```

```
# Define the file path in the Azure Blob Storage
file_path = "wasbs://<container-name>@<storage-account-name>.blob.core.windows.net/movies.csv"
```

```
# Read the file into a Spark DataFrame
df = spark.read.option("header", "true").csv(file_path)
```

```
# Show Data
df.show(5)
```
